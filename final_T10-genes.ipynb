From my shortlisted genes obtained from KEGG, I now want to validate that they're significant and not noisy genes

```python
upregulated_hits = {
    "AGAP2": "Endocytosis",
    "DNM3": "Endocytosis",
    "F2R": "Endocytosis",
    "HLA-B": "Endocytosis",
    "HSPA1L": "Endocytosis",
    "RET": "Endocytosis",
    "ATP6V0A2": "Lysosome",
    "AGA": "Lysosome",
    "AP3B2": "Lysosome",
    "ATP6AP1": "Lysosome",
    "CTSE": "Lysosome",
    "CTSF": "Lysosome",
    "CTSS": "Lysosome / Apoptosis",
    "ABCB1": "ADC Resistance/ABC-type transporter",
    "TUBA4A": "Apoptosis",
    "ETV1": "MMAE Resistance (up)",
    "KLHL4": "MMAE Resistance (up)",
    "KIAA1549L": "MMAE Resistance (down)",
    "UST": "MMAE Resistance (down)",
    "ANPEP": "EMT",
    "ADAM12": "EMT",
    "COL5A3": "EMT",
    "FZD8": "EMT",
    "IL32": "EMT",
    "IL6": "EMT",
    "MAGEE1": "EMT",
    "PMP22": "EMT",
    "SERPINE2": "EMT"
}

downregulated_hits = {
    "EHD3": "Endocytosis",
    "ASAH1": "Lysosome",
    "PSAPL1": "Lysosome",
    "GJA1": "EMT", 
    "PDZK1": "MMAE Resistance (up)",
    "PRR16": "MMAE Resistance (up)",
    "FAP": "EMT",
    "COL5A1": "EMT",
    "SPP1": "EMT",
    "ABCB11": "ABC-type transporter",
    "ABCG1": "ABC-type transporter",
    "GADD45B": "Apoptosis"
}

combined_hits = {**upregulated_hits, **downregulated_hits}
```

```python
combined_df = pd.read_csv(dir+'/data/combined-lines_summary.csv')
combined_df
```

```python
# Compact linear version
for gene in combined_hits.keys():
    count = 0
    
    if gene in combined_df['GeneSymbol'].values:
        # Check RT112
        row = combined_df[(combined_df['GeneSymbol'] == gene) & (combined_df['line'] == 'RT112')]
        if len(row) > 0 and row['padj'].iloc[0] < 0.05 and abs(row['log2FoldChange'].iloc[0]) > 1:
            count = count + 1
        
        # Check UMUC1
        row = combined_df[(combined_df['GeneSymbol'] == gene) & (combined_df['line'] == 'UMUC1')]
        if len(row) > 0 and row['padj'].iloc[0] < 0.05 and abs(row['log2FoldChange'].iloc[0]) > 1:
            count = count + 1
        
        # Check 647V
        row = combined_df[(combined_df['GeneSymbol'] == gene) & (combined_df['line'] == '647V')]
        if len(row) > 0 and row['padj'].iloc[0] < 0.05 and abs(row['log2FoldChange'].iloc[0]) > 1:
            count = count + 1
        
        print(f"{gene}: significant in {count}/3 cell lines")
    else:
        print(f"{gene}: not found in dataframe")
```
