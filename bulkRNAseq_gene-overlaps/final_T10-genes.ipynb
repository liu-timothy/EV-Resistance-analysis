```python
combined_df = pd.read_csv(dir+'/data/combined-lines_summary.csv')
combined_df
```

Check significance of upregulated hits
- p<0.05, log2FC>1

```python
for gene in upregulated_hits.keys():
    count = 0
    
    if gene in combined_df['GeneSymbol'].values:
        # Check RT112
        row = combined_df[(combined_df['GeneSymbol'] == gene) & (combined_df['line'] == 'RT112')]
        if len(row) > 0 and row['padj'].iloc[0] < 0.05 and row['log2FoldChange'].iloc[0] > 1:
            count = count + 1
        
        # Check UMUC1
        row = combined_df[(combined_df['GeneSymbol'] == gene) & (combined_df['line'] == 'UMUC1')]
        if len(row) > 0 and row['padj'].iloc[0] < 0.05 and row['log2FoldChange'].iloc[0] > 1:
            count = count + 1
        
        # Check 647V
        row = combined_df[(combined_df['GeneSymbol'] == gene) & (combined_df['line'] == '647V')]
        if len(row) > 0 and row['padj'].iloc[0] < 0.05 and row['log2FoldChange'].iloc[0] > 1:
            count = count + 1
        
        print(f"{gene}: significant in {count}/3 cell lines")
    else:
        print(f"{gene}: not found in dataframe")
    
    if count < 2:
        #print logFC and padj for each cell line
        print(f"{gene} details:")
        
        for line in ['RT112', 'UMUC1', '647V']:
            row = combined_df[(combined_df['GeneSymbol'] == gene) & (combined_df['line'] == line)]
            if len(row) > 0:
                print(f"  {line}: log2FC = {row['log2FoldChange'].iloc[0]:.3f}, padj = {row['padj'].iloc[0]:.3e}")
            else:
                print(f"  {line}: not found")
        print()
```
