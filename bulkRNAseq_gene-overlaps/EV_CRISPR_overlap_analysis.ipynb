```python
import numpy as np
import pandas as pd

from nbqol import path_to_git_root

dir = path_to_git_root() # automatically makes path to repo directory (wherever you cloned it)

print("Repository directory: ", dir)

# type(dir)
```

```python
'''
Preprocessing crispr ev data
'''

crispr_df = pd.read_csv(dir + '/data/EV_vs_PBS_DESeq2_results (Screen DESEQ2).txt', sep='\t')
crispr_df = crispr_df.drop(['sgID'], axis=1)
crispr_df = crispr_df[crispr_df['padj'] < 0.05]
crispr_df['line'] = 'CRISPR-screen'
crispr_df
```

```python
'''
Preprocessing rt112 data
'''

rt112 = pd.read_csv(dir + '/data/rp_vs_rr/rp_vs_rr_summary.txt', sep='\t')
rt112 = rt112[rt112['padj'] < 0.05] # p-val < 0.05
rt112 = rt112.rename(columns={'GeneSymbol': 'gene'})
#rt112['padj'].describe()
rt112['line'] = 'RT112'
rt112
```

Sort by ALL upregulated and ALL downregulated genes --> then find overlap

```python
'''
Sort by ALL upregulated and ALL downregulated genes --> then find overlap
- Requires log2FoldChange > +1 for upregulated and < -1 for downregulated
'''

rt112_upregulators = rt112.sort_values('log2FoldChange', ascending = False) #top upregulated genes in EV-res
rt112_upregulators = rt112_upregulators[rt112_upregulators['log2FoldChange'] > 1]

rt112_downregulators = rt112.sort_values('log2FoldChange', ascending = True) #top downregulated genes in EV-res
rt112_downregulators = rt112_downregulators[rt112_downregulators['log2FoldChange'] < -1]


# CRISPR DOWNREGULATORS ARE GENES UPREGULATED IN CRISPR SCREEN (log2FoldChange > 1)
crispr_downregulators = crispr_df.sort_values('log2FoldChange', ascending = False) #top upregulated genes in CRISPR
crispr_downregulators = crispr_downregulators[crispr_downregulators['log2FoldChange'] > 1]

# CRISPR UPREGULATORS ARE GENES DOWNREGULATED IN CRISPR SCREEN (log2FoldChange < -1)
crispr_upregulators = crispr_df.sort_values('log2FoldChange', ascending = True) #top downregulated genes in CRISPR
crispr_upregulators = crispr_upregulators[crispr_upregulators['log2FoldChange'] < -1]

print("Number of upregulated genes in RT112: ", len(rt112_upregulators))
print("Number of downregulated genes in RT112: ", len(rt112_downregulators))

print("Number of upregulating genes in CRISPR: ", len(crispr_upregulators))
print("Number of downregulating genes in CRISPR: ", len(crispr_downregulators))
```

```python
overlap = set(rt112_upregulators['gene'].tolist()) & set(crispr_upregulators['gene'].tolist())
print('Overlapping upregulated genes: ', overlap)

overlap = set(rt112_downregulators['gene'].tolist()) & set(crispr_downregulators['gene'].tolist())
print('Overlapping downregulated genes: ', overlap)
```

Sort by t100 upregulated and t100 downregulated genes --> then find overlap

```python
'''
Sort by t100 upregulated and t100 downregulated genes --> then find overlap
- Doesn't take into account log2FoldChange values
'''

rt112_top100 = rt112_upregulators.sort_values('log2FoldChange', ascending = False).head(100) #top 100 upregulated genes
rt112_bottom100 = rt112_downregulators.sort_values('log2FoldChange', ascending = True).head(100) #top 100 downregulated genes

crispr_top100 = crispr_upregulators.sort_values('log2FoldChange', ascending = False).head(100) #top 100 upregulated genes
crispr_bottom100 = crispr_downregulators.sort_values('log2FoldChange', ascending = True).head(100) #top 100 downregulated genes

overlap = set(rt112_top100['gene'].tolist()) & set(crispr_top100['gene'].tolist())
print('Overlapping upregulated genes: ', overlap)

overlap = set(rt112_bottom100['gene'].tolist()) & set(crispr_bottom100['gene'].tolist())
print('Overlapping downregulated genes: ', overlap)
```
